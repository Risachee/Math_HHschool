## Задание 2.1. Нормальная выборка
- сгенерировать 5000 значений нормального распределения, например со средним 40 и стандартным отклонением 8;
- построить гистограмму и нарисовать поверх неё теоретическую плотность;
- сравнить среднее и стандартное отклонение выборки с теми, что вы задали


```python
n, mu, sigma = 5000, 40, 8
sample = np.random.normal(loc=mu, scale=sigma, size=n)
sample_manual = np.linspace(mu - 4*sigma, mu + 4*sigma, 500)
sample_builtin = stats.norm.pdf(sample_manual, loc=mu, scale=sigma)

plt.figure(figsize=(9, 5))
plt.hist(sample, bins=40, density=True, alpha=0.6,
         color='pink', edgecolor='pink', label='Гистограмма выборки')
plt.plot(sample_manual, sample_builtin, 'r-', linewidth=2, label=f'плотность N{mu,sigma}')
plt.xlabel('Значение')
plt.ylabel('Плотность')
plt.title('Нормальная выборка n=5000')
plt.legend()
plt.grid(alpha=0.3)
plt.show()

print("Среднее выборки:", sample.mean())
print("Стандартное отклонение выборки:", sample.std())
print("Заданное среднее:", mu)
print("Заданное стандартное отклонение:", sigma)

print(f'\nРазница среднего: {mu - sample.mean():.4f}.\nРазница стандартного отклонения: {sigma - sample.std():.4f}')

```

## Задание 2.2. Биномиальная выборка
- сгенерировать 5000 значений биномиального распределения, например n = 1000, p = 0,04;
- построить гистограмму и нарисовать поверх неё теоретическую функцию вероятности;
- сравнить среднее и стандартное отклонение выборки с теоретическими значениями.


```python
n, p = 1000, 0.04
sample = np.random.binomial(n, p, size=5000)
k = np.arange(sample.min(), sample.max() + 1)

plt.figure(figsize=(9, 5))
plt.hist(sample, bins=np.arange(sample.min() - 0.5, sample.max() + 1.5), density=True, alpha=0.6,
         color='pink', edgecolor='pink', label='Гистограмма выборки')
plt.plot(k, stats.binom.pmf(k,n,p), 'o-', color='red', linewidth=2, label=f'Binomial{n,p}')
plt.xlabel('Значение')
plt.ylabel('Вероятность')
plt.title('Биномиальная выборка n=5000')
plt.legend()
plt.grid(alpha=0.3)
plt.show()

theoretical_mean = n * p
theoretical_std = np.sqrt(n * p * (1 - p))

print("Среднее выборки:", sample.mean())
print("Теоретическое среднее:", theoretical_mean)

print("Стандартное отклонение выборки:", sample.std())
print("Теоретическое стандартное отклонение:", theoretical_std)

print(f'\nРазница среднего: {theoretical_mean - sample.mean():.4f}')
print(f'Разница стандартного отклонения: {theoretical_std - sample.std():.4f}')
```

## Задание 2.3. ЦПТ на экспоненциальном распределении
- сгенерировать выборку из 500 значений экспоненциального распределения и посчитать её среднее;
- повторить это несколько тысяч раз;
- построить гистограмму полученных средних и в двух-трёх предложениях описать, что на ней видно и почему.


```python
sample_size = 500
n_simulations = 5000
scale = 2

means = []

for i in range(n_simulations):
    sample = np.random.exponential(scale=scale, size=sample_size)
    means.append(sample.mean())

plt.figure(figsize=(9, 5))

plt.hist(means, bins=40, density=True, alpha=0.6, color='pink', edgecolor='pink', label='Распределение средних')

plt.axvline(scale, color='red', linestyle='--', linewidth=2, label='Теоретическое среднее')

plt.xlabel('Среднее значение выборки')
plt.ylabel('Плотность')
plt.title('ЦПТ на экспоненциальном распределении')
plt.legend()
plt.grid(alpha=0.3)
plt.show()

print("Среднее полученных средних:", np.mean(means))
print("Теоретическое среднее:", scale)
print("Стандартное отклонение средних:", np.std(means))
print("Теоретическая ошибка среднего:", scale / np.sqrt(sample_size))

print("На гистограмме полученных средних наблюдается распределение, близкое к нормальному:" \
"оно имеет форму колокола и сосредоточено около значения 2, хотя исходное экспоненциальное распределение является асимметричным. ") 
print("Распределение средних становится приблизительно нормальным благодаря центральной предельной теореме. " \
"При этом разброс средних значительно меньше разброса исходных значений, поскольку стандартное отклонение средних уменьшается пропорционально квадратному корню из размера выборки.")
```

## Задание 3.2. Проверить симуляцией
- предположить, что кнопка ни на что не влияет и конверсия в обеих группах одинаковая;
- много раз смоделировать две такие группы по 5000 соискателей и посчитать разницу их
конверсий;
- построить гистограмму разниц и посчитать, как часто случайно получается 0,3 п.п. или больше;
- ответить: шум это или эффект.

```python
# Исходные данные
n = 5000
old_clicks = 155
new_clicks = 170
p = (old_clicks + new_clicks) / (2 * n)
observed_diff = new_clicks / n - old_clicks / n
n_simulations = 100000

old_sim = np.random.binomial(n, p, size=n_simulations)
new_sim = np.random.binomial(n, p, size=n_simulations)

diffs = new_sim / n - old_sim / n

p_one_sided = np.mean(diffs >= observed_diff)
p_two_sided = np.mean(np.abs(diffs) >= abs(observed_diff))

plt.figure(figsize=(9, 5))
plt.hist(diffs * 100, bins=50, density=True, alpha=0.6, color='pink', edgecolor='pink', label='Симулированные разницы')

plt.axvline(observed_diff * 100, color='red', linestyle='--', linewidth=2, label='Наблюдаемая разница (+0,3 п.п.)')

plt.axvline(-observed_diff * 100, color='red', linestyle=':', linewidth=1.5, label='Обратная разница (-0,3 п.п.)'
)

plt.xlabel('Разница конверсий, п.п.')
plt.ylabel('Плотность')
plt.title('Симуляция A/B-теста при отсутствии эффекта')
plt.legend()
plt.grid(alpha=0.3)
plt.show()

print("Общая конверсия:", p)
print("Наблюдаемая разница:", observed_diff * 100, "п.п.")
print("Односторонняя вероятность:", p_one_sided)
print("Двусторонняя вероятность:", p_two_sided)

print("Полученных данных недостаточно, чтобы утверждать, что новая кнопка улучшила конверсию. Наблюдаемая разница вполне может быть результатом шума.")
```

## Задание 3.3. Выбросы во времени сессий
Время сессий за день, в минутах: 3, 4, 4, 5, 5, 6, 6, 7, 8, 52.
- найти выбросы правилом трёх сигм;
- найти выбросы по перцентилям;
- сравнить результаты и объяснить, какой способ сработал лучше и почему.

<div class="alert alert-block alert-info">
    <b> Правило трёх сигм использует среднее и стандартное отклонение. Значение 52 само увеличивает оба показателя, из-за чего граница обнаружения выброса становится выше. </b>
    <b> Перцентили меньше зависят от величины экстремального значения. Поэтому в данном примере метод перцентилей сработал лучше. </b>
</div>

```python
sessions = np.array([3, 4, 4, 5, 5, 6, 6, 7, 8, 52])

mean = sessions.mean()
std = sessions.std()

lower = mean - 3 * std
upper = mean + 3 * std

outliers_3sigma = sessions[
    (sessions < lower) | (sessions > upper)
]

print("Среднее:", mean)
print("Стандартное отклонение:", std)
print("Нижняя граница:", lower)
print("Верхняя граница:", upper)
print("Выбросы:", outliers_3sigma)
```

```python
q1 = np.percentile(sessions, 25)
q3 = np.percentile(sessions, 75)

iqr = q3 - q1

lower_iqr = q1 - 1.5 * iqr
upper_iqr = q3 + 1.5 * iqr

outliers_iqr = sessions[
    (sessions < lower_iqr) | (sessions > upper_iqr)
]

print("Первый квартиль:", q1)
print("Третий квартиль:", q3)
print("IQR:", iqr)
print("Нижняя граница:", lower_iqr)
print("Верхняя граница:", upper_iqr)
print("Выбросы:", outliers_iqr)
```
